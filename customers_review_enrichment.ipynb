```python
# Simple, Rating-based sentiment enrichment

# -------------------------------------------------------------------
#
# IDEA:  we use the Rating column (1-5) that customers already gave us.
#  A rating already tells us how the customer felt, so we just convert it into details analysis
# 
#   1. SentimentScore   -> a number between -1 and 1
#   2. SentimentCategory-> Positive / Neutral / Negative
#   3. SentimentBucket  -> which range the score falls into
#
# -------------------------------------------------------------------

import pandas as pd

# STEP 1: Read the raw reviews CSV directly (replace SQL Server fetch)
customer_reviews_df = pd.read_csv('customers-review.csv')
# Expected columns: ReviewID, CustomerID, ProductID, ReviewDate, Rating, ReviewText


# STEP 2: Convert Rating (1-5) into a SentimentScore between -1 and 1
def calculate_sentiment(rating):
    # Rating 1 -> -1.0 | Rating 3 -> 0.0 | Rating 5 -> 1.0
    return (rating - 3) / 2



# STEP 3: Turn the score into a simple 3-way label
def categorize_sentiment(score):
    if score > 0.05:
        return 'Positive'
    elif score < -0.05:
        return 'Negative'
    else:
        return 'Neutral'


def sentiment_bucket(score):
    if score > 0.05:
        return '0.05 to 1.0 (Positive)'
    elif score < -0.05:
        return '-1.0 to -0.05 (Negative)'
    else:
        return '-0.05 to 0.05 (Neutral)'



# STEP 4: Apply all three functions row by row (using Rating column only)
customer_reviews_df['SentimentScore'] = customer_reviews_df['Rating'].apply(calculate_sentiment)
customer_reviews_df['SentimentCategory'] = customer_reviews_df['SentimentScore'].apply(categorize_sentiment)
customer_reviews_df['SentimentBucket'] = customer_reviews_df['SentimentScore'].apply(sentiment_bucket)

# STEP 5: Preview and save
print(customer_reviews_df.head())
customer_reviews_df.to_csv('fact_customer_reviews_with_sentiment.csv', index=False)
```
